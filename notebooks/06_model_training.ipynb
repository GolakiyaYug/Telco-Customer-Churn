## Model Training

```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
```

```python
X_train_processed = pd.read_csv(
    "../data/processed/X_train_processed.csv"
)

X_test_processed = pd.read_csv(
    "../data/processed/X_test_processed.csv"
)

y_train = pd.read_csv(
    "../data/processed/y_train.csv"
).squeeze()

y_test = pd.read_csv(
    "../data/processed/y_test.csv"
).squeeze()
```

```python
print("X_train shape:", X_train_processed.shape)
print("X_test shape :", X_test_processed.shape)

print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
logistic_model.fit(
    X_train_processed,
    y_train
)

print("Logistic Regression model trained successfully!")
```

```python
y_train_pred = logistic_model.predict(
    X_train_processed
)

print(y_train_pred[:10])
```

```python
y_test_pred = logistic_model.predict(
    X_test_processed
)

print(y_test_pred[:10])
```

```python
y_test_proba = logistic_model.predict_proba(
    X_test_processed
)[:, 1]

print(y_test_proba[:10])
```

```python
import joblib

joblib.dump(
    logistic_model,
    "../models/logistic_regression_model.pkl"
)

print("Logistic Regression model saved successfully!")
```

```python
import pandas as pd

X_train_processed = pd.read_csv(
    "../data/processed/X_train_processed.csv"
)

X_test_processed = pd.read_csv(
    "../data/processed/X_test_processed.csv"
)

y_train = pd.read_csv(
    "../data/processed/y_train.csv"
).squeeze()

y_test = pd.read_csv(
    "../data/processed/y_test.csv"
).squeeze()

print("Data loaded successfully!")
print("X_train:", X_train_processed.shape)
print("X_test :", X_test_processed.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

balanced_logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

print("Balanced Logistic Regression model created!")
```

```python
balanced_logistic_model.fit(
    X_train_processed,
    y_train
)

print("Balanced Logistic Regression trained successfully!")
```

```python
y_test_pred_balanced = balanced_logistic_model.predict(
    X_test_processed
)

y_test_proba_balanced = balanced_logistic_model.predict_proba(
    X_test_processed
)[:, 1]

print("Predictions generated successfully!")
```

```python
import joblib

joblib.dump(
    balanced_logistic_model,
    "../models/balanced_logistic_regression_model.pkl"
)

print("Balanced Logistic Regression model saved successfully!")
```
