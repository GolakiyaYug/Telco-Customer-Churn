```python
import pandas as pd
import numpy as np
```

```python
print("\nColumns:")
print(df.columns.tolist())
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())
```

```python
print("\nBasic Statistics:")
print(df.describe())
```

```python
print("\nStatistics including categorical columns:")
print(df.describe(include="all"))
```

```python
print("\nUnique values in each column:")

for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
print("\nNumber of unique values:")
print(df.nunique())
```

```python
print("\nValue Counts:")

for column in df.columns:
    print(f"\n========== {column} ==========")
    print(df[column].value_counts(dropna=False))
```

```python
print("\nTarget Distribution:")

print(df["Churn"].value_counts())

print("\nTarget Distribution Percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)
```

```python
print("\nConstant Features:")

constant_features = []

for column in df.columns:
    if df[column].nunique(dropna=False) <= 1:
        constant_features.append(column)

print(constant_features)
```

```python
print("\nNear-Constant Features:")

near_constant_features = []

for column in df.columns:

    value_percentage = (
        df[column].value_counts(normalize=True, dropna=False).iloc[0]
    )

    if value_percentage >= 0.70:
        near_constant_features.append(
            (column, value_percentage * 100)
        )

print(near_constant_features)
```

```python
for column in ['SeniorCitizen', 'Dependents', 'PhoneService', 'Churn']:
    print("\n", column)
    print(df[column].value_counts(normalize=True, dropna=False) * 100)
```

```python
print("\nSuspicious / Invalid Values:")

# Check blank strings
for column in df.columns:

    blank_count = (df[column].astype(str).str.strip() == "").sum()

    if blank_count > 0:
        print(f"{column}: {blank_count} blank values")
        
        
# Check " " values
print("\nWhitespace values:")
for column in df.columns:

    whitespace_count = (
        df[column].astype(str).str.strip() == ""
    ).sum()

    if whitespace_count > 0:
        print(column, whitespace_count)
```

```python
# Numerical Columns
numerical_columns = df.select_dtypes(include="number").columns.tolist()

# Categorical Columns
categorical_columns = df.select_dtypes(exclude="number").columns.tolist()

print("Numerical Columns:")
print(numerical_columns)
print("Total Numerical Columns:", len(numerical_columns))

print("\nCategorical Columns:")
print(categorical_columns)
print("Total Categorical Columns:", len(categorical_columns))
```

```python
print("\nInvalid TotalCharges values:")

total_charges_numeric = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

invalid_total_charges = total_charges_numeric.isna().sum()

print(
    "Invalid TotalCharges values:",
    invalid_total_charges
)
```

```python
print("\n" + "=" * 60)
print("DATA QUALITY SUMMARY")
print("=" * 60)

print("\nDataset Shape:")
print(df.shape)

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nColumns with Missing Values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nConstant Features:")
print(constant_features)

print("\nNear-Constant Features:")
print(near_constant_features)

print("\nTarget Distribution:")
print(df["Churn"].value_counts())

print("\nTarget Distribution Percentage:")
print(df["Churn"].value_counts(normalize=True) * 100)

print("\nMemory Usage:")
print(
    round(
        df.memory_usage(deep=True).sum() / (1024 ** 2),
        2
    ),
    "MB"
)

print("\nData Types:")
print(df.dtypes)

print("\nData Quality Check Completed!")
```
